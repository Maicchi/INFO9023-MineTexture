```python
import cv2 as cv2
import numpy as np

NAME = "steampunk_purple_hat.png"

# Load the image
img = cv2.imread(NAME)

width = 150
height = 150
dim = (width, height)

# Resize the image using INTER_AREA
resized_image = cv2.resize(img, dim, interpolation=cv2.INTER_AREA)

# Display or save the result
cv2.imwrite(NAME, resized_image)
```

```python
file_name = NAME

src = cv2.imread(file_name, 1)
gray = cv2.cvtColor(src, cv2.COLOR_BGR2GRAY)
blurred = cv2.GaussianBlur(gray, (5, 5), 0)

# 2. Threshold to get a binary mask of the object
# We use THRESH_BINARY_INV so the object is WHITE and background is BLACK
_, thresh = cv2.threshold(blurred, 245, 255, cv2.THRESH_BINARY_INV)

# 3. Find the contours
# RETR_EXTERNAL only gets the outermost boundary (ignoring internal holes)
contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# 4. Create an empty black mask the same size as the image
mask = np.zeros(gray.shape, dtype=np.uint8)

# 5. Draw the contours onto the mask and fill them with white (255)
# -1 means "fill the inside"
cv2.drawContours(mask, contours, -1, (255), thickness=cv2.FILLED)

# erosion
kernel = np.ones((3, 3), np.uint8)
mask = cv2.erode(mask, kernel, iterations=3)

# 6. Use this mask as your Alpha channel
b, g, r = cv2.split(src)
dst = cv2.merge([b, g, r, mask])

cv2.imwrite("steampunk_purple_hat_test.png", dst)
```

```python
src = cv2.imread("steampunk_purple_hat.png", 1)
image = np.array(src)
if image.shape[2] == 4:
    image = cv2.cvtColor(image, cv2.COLOR_RGBA2BGR)

mask = np.zeros(image.shape[:2], np.uint8)
backgroundModel = np.zeros((1, 65), np.float64)
foregroundModel = np.zeros((1, 65), np.float64)
height, width = image.shape[:2]
rect = (10, 10, width - 20, height - 20)

cv2.grabCut(
    image, mask, rect, backgroundModel, foregroundModel, 5, cv2.GC_INIT_WITH_RECT
)
mask2 = np.where((mask == 2) | (mask == 0), 0, 1).astype("uint8")
alpha = mask2 * 255
b, g, r = cv2.split(image)
result = cv2.merge([b, g, r, alpha])
cv2.imwrite("steampunk_purple_hat_test.png", result)
```
